```python
import os

# Change REPO-NAME to your actual repo folder name
REPO = r"C:\Users\phane\MSAI-Develop"

os.chdir(REPO)
print("Working folder:", os.getcwd())
print("Contents:", os.listdir())
```

```python
TEMPLATES = os.path.join(REPO, "templates")
os.makedirs(TEMPLATES, exist_ok=True)

html = """<!DOCTYPE html>
<html>
<head><title>Hello World</title></head>
<body>
  <h1>Hello, World!</h1>
  <p>Flask app running from my Anaconda / JupyterLab development environment.</p>
</body>
</html>
"""

with open(os.path.join(TEMPLATES, "index.html"), "w") as f:
    f.write(html)

print("Created:", os.path.join(TEMPLATES, "index.html"))
```

```python
from flask import Flask, render_template

app = Flask(__name__, template_folder=TEMPLATES)

@app.route("/")
def hello_world():
    return render_template("index.html")

print("Flask app created")
```
